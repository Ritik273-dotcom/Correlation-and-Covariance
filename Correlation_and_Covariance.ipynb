Q1. Define Covariance and explain how it differs from Correlation in terms of scale and interpretation.

Answer: Covariance vs. Correlation
Covariance is a statistical measure that indicates the direction of the linear relationship between two random variables.

Positive Covariance: Variables tend to move in the same direction.

Negative Covariance: Variables tend to move in opposite directions.

Zero Covariance: No linear relationship exists.

While covariance tells you if two variables are increasing or decreasing together, correlation provides a standardized metric that allows you to compare the strength of relationships across different datasets and units.

Q2. What does a positive, negative, and zero covariance indicate about the relationship between two variables?

Answer: Covariance indicates the direction of the linear relationship between two variables. Here is what the different signs signify:

Positive Covariance
Indicates a direct relationship. As one variable increases, the other tends to increase as well. Conversely, as one decreases, the other also tends to decrease.

Negative Covariance
Indicates an inverse relationship. As one variable increases, the other tends to decrease, and vice versa.

Zero Covariance
Indicates no linear relationship between the variables. This means the movement of one variable provides no information about the direction in which the other might move.

Q3. Discuss the limitations of covariance as a measure of relationship between two variables. Why is correlation
preferred in many cases?

Answer: The question asks for a discussion on the limitations of covariance and why correlation is often the preferred metric for assessing the relationship between two variables.

Limitations of Covariance

While covariance indicates the direction of a linear relationship (positive or negative), it has several significant drawbacks:

Scale Dependency: Covariance is sensitive to the units of measurement. For example, the covariance between height and weight will change drastically if you switch from centimeters and kilograms to inches and pounds.

Lack of Magnitude Interpretation: Because it is scale-dependent, the absolute value of covariance is difficult to interpret. A value of
10000 doesn't necessarily imply a stronger relationship than a value of
10; it might simply reflect larger measurement scales.

No Fixed Bounds: Covariance can range from - infinity to +infinity, making it impossible to judge the "strength" of a relationship at a glance.

Why Correlation is Preferred

Dimensionless: Correlation is a pure number with no units. This allows you to compare the relationships between completely different types of data (e.g., comparing "height vs. weight" to "income vs. education").

Standardized Range: Correlation is strictly bounded between -1 and +1.
+1: Perfect positive linear relationship.
-1: Perfect negative linear relationship.
0: No linear relationship.

Interpretable Strength: Because of its fixed scale, researchers can easily determine the strength of an association (e.g., 0.85 indicates a strong positive relationship, whereas 0.15 indicates a weak one).

Answer: Limitations of Covariance

Covariance indicates the direction of a linear relationship between two variables, but it has significant drawbacks as a descriptive measure:

Scale Dependency: The value of covariance depends entirely on the units of measurement. If you change a variable from meters to millimeters, the covariance will increase by a factor of 1,000, even though the underlying relationship hasn't changed.

Lack of Magnitude Interpretation: Because it isn't standardized, the absolute value of covariance is difficult to interpret. A "large" covariance doesn't necessarily mean a strong relationship; it might just reflect large scales in the data.

Sensitivity to Outliers: Like the mean, covariance is highly sensitive to extreme values, which can distort the perceived relationship.

Q4. Explain the difference between Pearson’s correlation coefficient and Spearman’s rank correlation
coefficient. When would you prefer to use Spearman’s correlation?

Answer: The primary difference between these two coefficients lies in the type of relationship they measure and the type of data they require.

image.png

When to prefer Spearman's Correlation

You should choose Spearman’s rank correlation in the following scenarios:

Non-linear Relationships: When the relationship is monotonic (as one variable increases, the other consistently increases or decreases) but not a straight line (e.g., exponential growth).

Ordinal Data: When your data consists of rankings or scales (e.g., finishing positions in a race or Likert scales like "satisfied" to "unsatisfied").

Outliers: When your dataset contains significant outliers that would skew a Pearson calculation.

Non-normal Distribution: When your data violates the assumption of normality required for parametric tests.

Why or why not?

A high correlation simply shows that two variables move together; it does not prove that one causes the other. This is due to several factors:

Confounding Variables: A third, unmeasured variable (a "lurking" variable) might be influencing both
